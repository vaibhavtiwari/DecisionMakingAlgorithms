## Exact Inference by Enumeration

## 1. Objective

The objective of **Inference by Enumeration** is to compute the exact posterior probability of the query variables given the observed evidence by explicitly summing over all possible assignments of the hidden variables.

Although computationally expensive, enumeration forms the foundation of many exact inference algorithms.

## 2. Mathematical Background

Given

- Query variables: $Q$
- Evidence variables: $E=e$
- Hidden variables: $H$

our objective is to compute: $$P(Q∣e)$$.

Using Bayes' Theorem, $$P(Q|e) = \frac{P(Q,e)}{P(e)}$$

Since, the hidden variables are unknown, $$P(Q,e) = \sum_H P(Q,H,e)$$

Therefore, $$P(Q|e) = \alpha\sum_H P(Q,H,e)$$

where, $\alpha = \frac{1}{P(e)}$ is the normalization constant

For a Bayesian Network, the joint distribution factorizes as $$P(X_1,X_2,...,X_n)= \prod_{i=1}^{n} P(X_i | Pa(X_i)) $$
which gives $$P(Q|e) = \alpha\sum_H \prod_{i=1}^{n} P(X_i | Pa(X_i))$$

This equation is evaluated directly by the enumeration algorithm.

## 3. Algorithm

## Idea

Rather than trying to derive the posterior directly, the algorithm considers every possible assignment of the hidden variables, computes the corresponding joint probability, sums all contributions, and finally normalizes the result.

## Algorithm

ENUMERATION-ASK(Query, Evidence)

for each possible value q of the query variable

    total = 0

    for each assignment h of the hidden variables

        total += P(q, h, evidence)

normalize(total)

return posterior

## 4. Python Implementation

### Generate Hidden Variable Assignments

````python
from itertools import product
from typing import Any


def enumerate_assignments(
    hidden_domains: dict[str, list[Any]],
):
    """
    Generate all possible assignments of hidden variables.

    Parameters
    ----------
    hidden_domains : dict
        Mapping from variable names to their domains.

    Yields
    ------
    dict
        One assignment of all hidden variables.
    """

    variables = list(hidden_domains.keys())
    domains = [hidden_domains[var] for var in variables]

    for values in product(*domains):
        yield dict(zip(variables, values))
````

## 5. Worked Example

Consider the following Bayesian Network.

```text
Cloudy
   |
   v
Rain
 /   \
v     v
WetGrass   Traffic
```

Suppose we observe

$$
\text{WetGrass} = \text{True}
$$

and want to compute the posterior probability

$$
P(\text{Rain}\mid\text{WetGrass}=\text{True}).
$$

### Step 1: Identify the Variables

- **Query Variable**

$$
Q = \text{Rain}
$$

- **Evidence Variable**

$$
E = \{\text{WetGrass}=\text{True}\}
$$

- **Hidden Variable**

$$
H = \{\text{Cloudy}\}
$$

---

### Step 2: Apply the General Inference Equation

Using the inference equation,

$$
P(Q\mid e)
=
\alpha
\sum_H
P(Q,H,e),
$$

we obtain

$$
P(\text{Rain}\mid\text{WetGrass})
=
\alpha
\sum_{\text{Cloudy}}
P(\text{Rain},\text{Cloudy},\text{WetGrass}).
$$

Since **Cloudy** is a binary variable, the summation consists of two terms:

$$
P(\text{Rain},\text{WetGrass})
=
P(\text{Rain},\text{Cloudy=True},\text{WetGrass})
+
P(\text{Rain},\text{Cloudy=False},\text{WetGrass}).
$$

Similarly,

$$
P(\neg\text{Rain},\text{WetGrass})
=
P(\neg\text{Rain},\text{Cloudy=True},\text{WetGrass})
+
P(\neg\text{Rain},\text{Cloudy=False},\text{WetGrass}).
$$

---

### Step 3: Normalize

Finally, the posterior distribution is obtained by normalization,

$$
P(\text{Rain}\mid\text{WetGrass})
=
\frac{
P(\text{Rain},\text{WetGrass})
}{
P(\text{Rain},\text{WetGrass})
+
P(\neg\text{Rain},\text{WetGrass})
}.
$$

---

### Interpretation

Inference by enumeration computes the posterior probability by explicitly evaluating every possible assignment of the hidden variables.

In this example, only one hidden variable (**Cloudy**) exists, so only two assignments need to be evaluated.

| Cloudy | Contribution |
|--------|--------------|
| True | $P(\text{Rain},\text{Cloudy=True},\text{WetGrass})$ |
| False | $P(\text{Rain},\text{Cloudy=False},\text{WetGrass})$ |

For larger Bayesian Networks, the number of hidden variables increases, causing the number of possible assignments to grow exponentially. This exponential growth is the primary limitation of inference by enumeration and motivates more efficient algorithms such as **Variable Elimination**.

## 6. Complexity Analysis

Let

- $n$ be the number of hidden variables,
- $d$ be the maximum domain size.

The number of assignments is $d^n$

Therefore, 


| Complexity | Value |
|:---:|:---:|
| Time | $O(d^n)$ |
| Space | $O(n)$ |

## 7. Key Takeaways

1. Enumeration computes the exact posterior distribution.
2. Every assignment of the hidden variables is explicitly evaluated.
3. Hidden variables are marginalized by summing over all possible values.
4. The algorithm is conceptually simple but computationally expensive.
5. Enumeration forms the basis of more efficient exact inference algorithms.
