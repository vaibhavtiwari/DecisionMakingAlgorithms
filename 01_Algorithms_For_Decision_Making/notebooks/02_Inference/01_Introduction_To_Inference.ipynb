# Inference in Probabilistic Models

## Introduction

In the previous chapter, we studied **probabilistic representations**. We learned how uncertainty can be represented using probability distributions, factors, and Bayesian networks. These representations allow us to model relationships between random variables in a compact and mathematically consistent manner.

However, building a probabilistic model is only the first step.

The real purpose of such a model is to answer questions about the world.

For example,

- Is there a pedestrian hidden behind a parked vehicle?
- Is a patient likely to have a particular disease?
- Where is a mobile robot located inside a building?
- Will a machine fail within the next hour?

In all these situations, we do not directly observe the quantity of interest. Instead, we observe related evidence and use the probabilistic model to estimate the unknown variable.

This process is known as **probabilistic inference**.

## What is Inference?

Inference is the process of estimating unknown quantities from observed evidence using a probabilistic model.

Suppose we already have a probabilistic model describing relationships between variables.

Instead of asking, **"How do we represent uncertainty?"**, we now ask **"Given some observations, what can we conclude about the unknown variables?"**

Mathematically, inference computes probabilities of the form, $$P(Unknown Variables | Observed Evidence)$$

This conditional probability represents our updated belief after incorporating new observations.

## Representation vs Inference

It is useful to distinguish between these two concepts.

|           Representation          |              Inference             |
|:---------------------------------:|:----------------------------------:|
| Describes the probabilistic model | Uses the model to answer questions |
| Defines probability distribution  | Computes conditional probabilities |
| Stores knowledge                  | Reasons using knowledge            |
| Constructed before observations   | Performed after observations       |

Think of it this way:

**Representation** builds the map.

**Inference** uses the map to find the destination.

## Types of Variables

Almost every inference problem consists of three categories of variables.

- **Query Variables ($Q$):** The variables whose probabilities we want to compute.

- **Evidence Variables ($E$):** Variables whose values are already observed.

- **Hidden Variables ($H$):** Variables that are neither queried nor observed but still influence the computation.**These variables must usually be eliminated during inference.**
  

## General Inference Problem

Suppose the probabilistic model contains variables

$$X=\{X_1,X_2,…,X_n\}$$

We divide them into three disjoint sets

- Query variables $Q$
- Evidence variables $E$
- Hidden variables $H$

such that, $X=Q \cup E \cup H$.

Our goal is to compute, $P(Q∣E=e)$

where $e$ denotes the observed evidence.

## Why Hidden Variables Exist

In real-world problems, we rarely observe every variable.

Examples include

| Application | Hidden Variable | Observed Evidence |
|:---:|:---:|---|
| Medical diagnosis | Disease | Symptoms |
| Autonomous driving | Pedestrian intention | Camera, LiDAR |
| Robot localization | Robot position | GPS, IMU, LiDAR |
| Speech recognition | Spoken words | Audio signal |

Inference estimates the hidden variables using the available evidence.

## Mathematical Objective

Using Bayes' rule, $$P(Q|e) = \frac{P(Q,e)}{P(e)}$$

The denominator $P(e)$ acts as a normalization constant.

The difficulty lies in calculating the numerator, $P(Q,e)$ because the hidden variables are still unknown.

Therefore, $$P(Q,e) = \sum_H P(Q,H,e)$$

Substituting, $$P(Q|e) = \frac{\sum_H P(Q,H,e)}{P(e)}$$

Since, $$P(e) = \sum_Q \sum_H P(Q,H,e)$$

The inference equation becomes: $$P(Q|e) = \frac{\sum_H P(Q,H,e)}{\sum_Q \sum_H P(Q,H,e)}$$



## Introducing the Normalization Constant

To simplify notation, $$\alpha = \frac{1}{P(e)}$$

Therefore, $$P(Q,e) = \alpha\sum_H P(Q,H,e)$$

This compact equation appears repeatedly throughout probabilistic inference algorithms.

## Why is Inference Difficult?

Suppose a probabilistic model contains 100 binary variables.

The joint probability distribution contains $2^{100}$ entries.

this is approximately $1.27$ x $10^{30}$ probabilities.

Clearly, storing and summing over such a distribution is computationally infeasible.

Instead, modern inference algorithms exploit

- conditional independence,
- graph structure,
- factorization of the joint distribution.

These ideas dramatically reduce computational complexity.

## Exact and Approximate Inference

Inference algorithms can broadly be divided into two categories.

- **Exact Inference**

Computes the exact posterior probability.

Examples include

- Enumeration
- Variable Elimination
- Belief Propagation
- Junction Tree Algorithm
- Approximate Inference

Produces an approximation when exact computation becomes too expensive.

Examples include

- Monte Carlo Sampling
- Importance Sampling
- Gibbs Sampling
- Particle Filtering
- Variational Inference

## Intuition

**Representation** tells us how the world is modeled.

**Inference** tells us how the model answers questions.

## Research Perspective

In robotics and autonomous driving, sensors never provide a complete description of the environment. Cameras may be occluded, LiDAR measurements may be sparse, GPS may be noisy, and pedestrians may behave unpredictably. Consequently, intelligent systems must continuously infer hidden aspects of the world—such as object intentions, vehicle localization, or environmental state—from partial and uncertain observations. Probabilistic inference provides the mathematical foundation for this reasoning process and serves as the backbone of state estimation, localization, mapping, tracking, planning, and decision making in modern autonomous systems.
