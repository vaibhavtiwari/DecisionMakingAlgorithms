# Bayesian Network Scoring

## Motivation

In parameter learning, we assumed that the structure of the Bayesian network was already known.

For example, suppose we were given the structure

$$
A \rightarrow B,
\qquad
C.
$$

The structure tells us that the joint distribution factorizes as

$$
P(A,B,C)=
P(A)P(B\mid A)P(C).
$$

The remaining problem was to learn the parameters of these distributions from data.

Therefore, parameter learning can be viewed as

$$
\boxed{
\text{Known Structure }G
+
\text{Data }D
\rightarrow
\text{Learn Parameters }\theta
}
$$

---

## From Parameter Learning to Structure Learning

Now we remove the assumption that the graph structure is known.

Suppose we observe a dataset containing measurements of three variables

$$
A,\quad B,\quad C.
$$

Several Bayesian network structures could potentially explain the same data:

$$
A \rightarrow B,
\qquad
C
$$

or

$$
A \qquad B \qquad C
$$

or a more complex network containing several dependencies.

The question is therefore no longer only

> What parameters best describe the data?

We must also ask

> Which graph structure best describes the data?

Structure learning can therefore be viewed as

$$
\boxed{
\text{Data }D
\rightarrow
\text{Learn Graph Structure }G
}
$$

---

## The Structure Learning Problem

We would like to compare candidate graphs according to how well they explain the observed data.

Using a Bayesian approach, we consider

$$
P(G\mid D),
$$

the posterior probability of a graph structure given the data.

Using Bayes' rule,

$$
P(G\mid D)
\propto
P(G)P(D\mid G).
$$

Here,

- $P(G)$ is the prior probability of the graph,
- $P(D\mid G)$ measures how well the graph explains the observed data.

Therefore, structure learning can be formulated as

$$
\boxed{G^*=
\arg\max_G P(G\mid D)
}
$$

The first problem we need to solve is therefore:

$$
\boxed{
\text{Given a candidate graph }G
\text{ and dataset }D,
\text{ how do we score }G?
}
$$

## Marginalizing the Network Parameters

We want to evaluate a candidate graph using

$$
P(G\mid D)
\propto
P(G)P(D\mid G).
$$

However, a Bayesian network is not completely specified by its graph structure alone.

For example, consider

$$
A \rightarrow B.
$$

The graph tells us that

$$
P(A,B)=
P(A)P(B\mid A),
$$

but we still need parameters such as

$$
P(A=1),
$$

$$
P(B=1\mid A=0),
$$

and

$$
P(B=1\mid A=1).
$$

Let all of these network parameters be represented by

$$
\theta.
$$

Therefore, the likelihood of the data for a particular set of parameters is

$$
P(D\mid\theta,G).
$$

But during structure learning, we do not know the correct value of $\theta$.

---

### Bayesian Treatment of the Unknown Parameters

Instead of selecting one particular value of $\theta$, we consider all possible parameter values.

We place a prior distribution over the parameters,

$$
p(\theta\mid G),
$$

and marginalize $\theta$:

$$
\boxed{
P(D\mid G)=
\int
P(D\mid\theta,G)
p(\theta\mid G)
\,d\theta
}
$$

Substituting this into the graph posterior gives

$$
\boxed{
P(G\mid D)
\propto
P(G)
\int
P(D\mid\theta,G)
p(\theta\mid G)
\,d\theta
}
$$

This quantity allows us to evaluate the graph itself without committing to one particular parameter estimate.

---

## Connection to Bayesian Parameter Learning

This is the same Bayesian idea used in the previous chapter.

In Bayesian parameter learning, we retained uncertainty over the parameter:

$$
P(\theta\mid D).
$$

For posterior prediction, we marginalized the unknown parameter:

$$
P(x_{\text{new}}\mid D)=
\int
P(x_{\text{new}}\mid\theta)
P(\theta\mid D)
\,d\theta.
$$

Structure learning uses the same principle:

$$
\boxed{
\text{Unknown quantity}
\rightarrow
\text{represent uncertainty}
\rightarrow
\text{marginalize it}
}
$$

Here, the unknown quantity being marginalized is the network parameter $\theta$ so that we can evaluate the graph $G$ itself.

## Marginal Likelihood for a Binary Variable

Before deriving the score for an entire Bayesian network, consider the simplest possible model:

$$
X\sim\operatorname{Bernoulli}(\theta).
$$

Suppose the dataset contains

$$
N_1
$$

observations where $X=1$ and

$$
N_0
$$

observations where $X=0$.

The likelihood is

$$
P(D\mid\theta)=
\theta^{N_1}(1-\theta)^{N_0}.
$$

As in Bayesian parameter learning, place a Beta prior on $\theta$:

$$
\theta\sim\operatorname{Beta}(\alpha,\beta).
$$

Therefore,

$$
p(\theta)=
\frac{1}{B(\alpha,\beta)}
\theta^{\alpha-1}
(1-\theta)^{\beta-1}.
$$

---

### Marginalizing the Parameter

To evaluate the model without choosing a particular value of $\theta$, we integrate over all possible values:

$$
P(D)=
\int_0^1
P(D\mid\theta)
p(\theta)
\,d\theta.
$$

Substituting the likelihood and prior,

$$
P(D)=
\int_0^1
\theta^{N_1}
(1-\theta)^{N_0}
\frac{1}{B(\alpha,\beta)}
\theta^{\alpha-1}
(1-\theta)^{\beta-1}
\,d\theta.
$$

Combining the powers gives

$$
P(D)=
\frac{1}{B(\alpha,\beta)}
\int_0^1
\theta^{\alpha+N_1-1}
(1-\theta)^{\beta+N_0-1}
\,d\theta.
$$

The integral is itself a Beta function:

$$
B(\alpha+N_1,\beta+N_0).
$$

Therefore,

$$
\boxed{P(D)=
\frac{
B(\alpha+N_1,\beta+N_0)
}{
B(\alpha,\beta)
}
}
$$

This is the marginal likelihood of the observed data.

---

### Connection to the Gamma Function

The Beta function can be written as

$$
B(a,b)=
\frac{
\Gamma(a)\Gamma(b)
}{
\Gamma(a+b)
}.
$$

Therefore,

$$
P(D)=
\frac{
\Gamma(\alpha+\beta)
}{
\Gamma(\alpha+\beta+N_0+N_1)
}
\frac{
\Gamma(\alpha+N_1)
}{
\Gamma(\alpha)
}
\frac{
\Gamma(\beta+N_0)
}{
\Gamma(\beta)
}.
$$

This Gamma-function form is the same pattern that appears in the Bayesian network score.
